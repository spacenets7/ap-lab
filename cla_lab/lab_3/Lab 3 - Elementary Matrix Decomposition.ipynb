```python
import numpy as np
```

```python
def elementary_decomposition(A):
    A = np.array(A, dtype=float)
    n = A.shape[0]
    
    # Store inverses of elementary matrices
    elementary_matrices = []
    
    # Step 1: Scale first row
    if A[0,0] != 1:
        E1_inv = np.eye(n)
        E1_inv[0,0] = A[0,0]
        elementary_matrices.append(E1_inv)
        A[0] = A[0] / A[0,0]
    
    # Step 2: Eliminate below pivot
    if A[1,0] != 0:
        factor = A[1,0]
        E2_inv = np.eye(n)
        E2_inv[1,0] = factor
        elementary_matrices.append(E2_inv)
        A[1] = A[1] - factor * A[0]
    
    # Step 3: Scale second row
    if A[1,1] != 1:
        E3_inv = np.eye(n)
        E3_inv[1,1] = A[1,1]
        elementary_matrices.append(E3_inv)
        A[1] = A[1] / A[1,1]
    
    # Step 4: Eliminate above pivot
    if A[0,1] != 0:
        factor = A[0,1]
        E4_inv = np.eye(n)
        E4_inv[0,1] = factor
        elementary_matrices.append(E4_inv)
        A[0] = A[0] - factor * A[1]
    
    return elementary_matrices
```

```python
# Example matrix
A = [[4, 3],[2, 1]]
matrices = elementary_decomposition(A)
```

```python
# Verify the product
product = np.eye(2)
for M in matrices:
    product = product @ M
```

```python
print("Elementary matrices (in order):")
for i, M in enumerate(matrices, start=1):
    print(f"E{i}^-1:\n{M}\n")

print("Product of elementary matrices:")
print(product)
```

# EXERCISE

The digit root of a number is a single-digit number obtained by repeatedly summing the digits of a number until a single-digit result is achieved. Form an elementary decomposition of the matrix $$A =
\begin{pmatrix}
5 & K \\
2 & K
\end{pmatrix}
$$

where K is the digit root of your roll number. Illustrate the effect of multiplying the vector $(1,0)$ by $A$ geometrically using the elementary transformations. Verify your answer algebraically.

```python
roll_number = 59

def digit_root(n):
    n = abs(n)
    while n >= 10:
        n = sum(int(d) for d in str(n))
    return n

K = digit_root(roll_number)
print("K (digit root of roll number) =", K)


A_ex = [[5, K], [2, K]]
print("A =")
print(np.array(A_ex))

matrices_ex = elementary_decomposition(A_ex)

print("Elementary matrices (in order):")
for i, M in enumerate(matrices_ex, start=1):
    print(f"E{i}^-1:\n{M}\n")


product_ex = np.eye(2)
for M in matrices_ex:
    product_ex = product_ex @ M

print("Product of elementary matrices:")
print(product_ex)


v = np.array([1.0, 0.0])
print("v0 =", v)

for i, M in enumerate(reversed(matrices_ex), start=1):
    label = len(matrices_ex) - i + 1
    v = M @ v
    print(f"after applying E{label}^-1: v = {v}")

print("\nFinal transformed vector A @ (1,0):", v)


direct = np.array(A_ex, dtype=float) @ np.array([1.0, 0.0])
print("Direct A @ (1,0) =", direct)
print("Matches step-by-step result:", np.allclose(direct, v))
```

```python
import matplotlib.pyplot as plt
from collections import defaultdict


v = np.array([1.0, 0.0])
points = [v.copy()]
labels = ["v0 = (1,0)"]

for i, M in enumerate(reversed(matrices_ex), start=1):
    label = len(matrices_ex) - i + 1
    v = M @ v
    points.append(v.copy())
    labels.append(f"after E{label}^-1")

points = np.array(points)

fig, ax = plt.subplots(figsize=(6, 6))
colors = plt.cm.viridis(np.linspace(0, 1, len(points)))


coord_groups = defaultdict(list)
for i, pt in enumerate(points):
    coord_groups[tuple(np.round(pt, 6))].append(i)

for i, (pt, lab, col) in enumerate(zip(points, labels, colors)):
    ax.arrow(0, 0, pt[0], pt[1],
              head_width=0.15, length_includes_head=True,
              color=col, linewidth=2, alpha=0.85,
              zorder=i)  

 
    group = coord_groups[tuple(np.round(pt, 6))]
    slot = group.index(i)
    text_offset = 0.25 + 0.35 * slot   
    ax.annotate(lab, xy=(pt[0], pt[1]),
                xytext=(pt[0] + 0.2, pt[1] + text_offset),
                fontsize=9, color=col,
                arrowprops=dict(arrowstyle='-', color=col, alpha=0.5, lw=0.7))

max_val = points.max() + 1
ax.set_xlim(-1, max_val)
ax.set_ylim(-1, max_val)
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)
ax.set_aspect('equal')
ax.grid(True, linestyle='--', alpha=0.4)
ax.set_title("Geometric effect of elementary transformations on (1,0)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend(labels, loc='upper left', fontsize=8, framealpha=0.9)

plt.tight_layout()
plt.savefig("elementary_transform_geometry.png", dpi=150)
plt.show()
```
